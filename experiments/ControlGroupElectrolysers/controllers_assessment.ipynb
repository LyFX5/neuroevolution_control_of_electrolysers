```python
import pickle
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
file_assessment_df = open('controllers_assessment/mpc_results_5elec.pkl', 'rb')
assessment_df_mpc_5 = pickle.load(file_assessment_df)
file_assessment_df.close()
```

```python
print("±")
```

```python
days_with_low_unclaimed_energy = assessment_df_mpc_5[(assessment_df_mpc_5.deterioration_mean == 0.0)].index
```

```python
def get_assessment_df(path):
    file_assessment_df = open(path, 'rb')
    assessment_df = pickle.load(file_assessment_df)
    file_assessment_df.close()
    assessment_df = assessment_df.drop(days_with_low_unclaimed_energy)
    assessment_df["total_deterioration_wrt_unclaimed_energy"] = (5 * assessment_df["deterioration_mean"]) / assessment_df["unclaimed_energy"]
    assessment_df["total_deterioration_wrt_consumed_energy"] = (5 * assessment_df["deterioration_mean"]) / assessment_df["consumed_energy"]
    metrics_mean = {"day": [], "RMSE": [], "MSE": [], "MAE": [], "SMAPE": [], 
                       "unclaimed_energy": [], "consumed_energy": [], "energy_loss": [], 
                       "deterioration": [], "deterioration_max": [], "deterioration_mean": [], "deterioration_std": []}
    metrics_mean["day"] = "avg"
    metrics_mean["RMSE"] = [np.mean(assessment_df["RMSE"]), np.std(assessment_df["RMSE"])]
    metrics_mean["MSE"] = [np.mean(assessment_df["MSE"]), np.std(assessment_df["MSE"])] 
    metrics_mean["MAE"] = [np.mean(assessment_df["MAE"]), np.std(assessment_df["MAE"])] 
    metrics_mean["SMAPE"] = [np.mean(assessment_df["SMAPE"]), np.std(assessment_df["SMAPE"])] 
    metrics_mean["unclaimed_energy"] = [np.mean(assessment_df["unclaimed_energy"]), np.std(assessment_df["unclaimed_energy"])]
    metrics_mean["consumed_energy"] = [np.mean(assessment_df["consumed_energy"]), np.std(assessment_df["consumed_energy"])]
    metrics_mean["energy_loss"] = [np.mean(assessment_df["energy_loss"]), np.std(assessment_df["energy_loss"])]
    metrics_mean["deterioration"] = "avg" # [np.mean(assessment_df_mpc_5[""]), np.std(assessment_df_mpc_5[""])] group_deterioration
    metrics_mean["deterioration_max"] = [np.mean(assessment_df["deterioration_max"]), np.std(assessment_df["deterioration_max"])]
    metrics_mean["deterioration_mean"] = [np.mean(assessment_df["deterioration_mean"]), np.std(assessment_df["deterioration_mean"])]
    metrics_mean["deterioration_std"] = [np.mean(assessment_df["deterioration_std"]), np.std(assessment_df["deterioration_std"])]
    metrics_mean["total_deterioration_wrt_unclaimed_energy"] = [np.mean(assessment_df["total_deterioration_wrt_unclaimed_energy"]), np.std(assessment_df["total_deterioration_wrt_unclaimed_energy"])]
    metrics_mean["total_deterioration_wrt_consumed_energy"] = [np.mean(assessment_df["total_deterioration_wrt_consumed_energy"]), np.std(assessment_df["total_deterioration_wrt_consumed_energy"])] 
    assessment_df = assessment_df.append([metrics_mean], ignore_index=True)
    return assessment_df
```

```python
df_heuristic_5 = get_assessment_df("controllers_assessment/heuristic_results_5elec.pkl")
df_mpc_5 = get_assessment_df("controllers_assessment/mpc_results_5elec.pkl")
df_neuro_5 = get_assessment_df("controllers_assessment/neuro_results_5elec.pkl")
df_heuristic_2 = get_assessment_df("controllers_assessment/heuristic_results_2elec.pkl")
df_mpc_2 = get_assessment_df("controllers_assessment/mpc_results_2elec.pkl")
df_neuro_2 = get_assessment_df("controllers_assessment/neuro_results_2elec.pkl")
```

```python
df_mean_metrics = pd.DataFrame({"heuristic_2": [], "mpc_2": [], "neuro_2": [], "heuristic_5": [], "mpc_5": [], "neuro_5": []})
```

```python
df_mean_metrics["heuristic_2"] = df_heuristic_2.iloc[-1]
df_mean_metrics["mpc_2"] = df_mpc_2.iloc[-1]
df_mean_metrics["neuro_2"] = df_neuro_2.iloc[-1]
df_mean_metrics["heuristic_5"] = df_heuristic_5.iloc[-1]
df_mean_metrics["mpc_5"] = df_mpc_5.iloc[-1]
df_mean_metrics["neuro_5"] = df_neuro_5.iloc[-1]
```
